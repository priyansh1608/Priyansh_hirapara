```python
"""
4.  Given a list of 10 user reviews for a restaurant (as text), write a Python function
    that simulates batch prediction by categorizing each review as 'positive' or 
    'negative' based on whether it contains the word 'good'. Print the results for all
    reviews.<br><br><em><strong>Constraint:</strong> Do not use any external ML 
    libraries; just use basic string operations.</em>
"""

def batch_prediction(reviews):
    results = []

    for review in reviews:
        if "good" in review.lower():
            results.append("positive")
        else:
            results.append("negative")

    return results


reviews = [
    "The food was good and tasty",
    "The service was very slow",
    "Good food and excellent service",
    "The restaurant was dirty",
    "The pizza was good",
    "The food was bad",
    "Very good experience",
    "The staff was rude",
    "The taste was good",
    "I did not enjoy the food"
]

results = batch_prediction(reviews)

for review, result in zip(reviews, results):
    print("Review:", review)
    print("Prediction:", result)
    print()
```
