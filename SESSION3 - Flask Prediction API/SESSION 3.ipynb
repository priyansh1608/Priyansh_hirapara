```python
# Q1
"""
1.  Install Flask in a new Python virtual environment and create a file called app.py that displays 'Welcome to the
    Prediction API' at the root URL (/).
"""
from flask import Flask
import threading

app = Flask(__name__)

@app.route("/")
def home():
    return "Welcome to the Prediction API"

# Run Flask in background
thread = threading.Thread(
    target=lambda: app.run(
        debug=False,
        use_reloader=False,
        port=5000
    )
)

thread.daemon = True
thread.start()

print("Flask API started successfully!")
print("Open: http://127.0.0.1:5000/")
```

```python
#Q2

"""
2.  Build a simple REST API in Flask called /predict-price that accepts a POST request with JSON input containing
    'base_price' and 'discount' fields, and returns the final price after applying the discount.
"""
from flask import Flask, request, jsonify
import threading

app = Flask(__name__)

@app.route("/predict-price", methods=["POST"])
def predict_price():
    data = request.get_json()

    base_price = data["base_price"]
    discount = data["discount"]

    final_price = base_price - (base_price * discount / 100)

    return jsonify({
        "base_price": base_price,
        "discount": discount,
        "final_price": final_price
    })

thread = threading.Thread(
    target=lambda: app.run(
        debug=False,
        use_reloader=False,
        port=5001
    )
)

thread.daemon = True
thread.start()

print("Price Prediction API started!")
print("URL: http://127.0.0.1:5001/predict-price")
```

```python
import requests

url = "http://127.0.0.1:5001/predict-price"

data = {
    "base_price": 1000,
    "discount": 20
}

response = requests.post(url, json=data)

print(response.json())
```

```python
#Q3

"""
3.  Save a trained scikit-learn LinearRegression model (for example, predicting delivery time based on distance and 
    order size) to a file using joblib, then load this model in your Flask app and use it to make predictions when the
    /predict-delivery endpoint receives a POST request with JSON input.
"""


from sklearn.linear_model import LinearRegression
import joblib

# Training data
X = [
    [1, 1],
    [2, 1],
    [3, 2],
    [4, 2],
    [5, 3],
    [6, 3],
    [7, 4],
    [8, 4]
]

# Delivery time
y = [15, 18, 22, 25, 30, 34, 38, 42]

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Save model
joblib.dump(model, "delivery_model.pkl")

print("Model saved successfully!")
```

```python
import joblib

model = joblib.load("delivery_model.pkl")

print("Model loaded successfully!")
```

```python
distance = 5
order_size = 3

prediction = model.predict([[distance, order_size]])

print("Predicted delivery time:", round(prediction[0], 2), "minutes")
```

```python
#Q4

"""
4.  Update your Flask API so that it returns a JSON response with both the predicted value and a custom message, 
    similar to how Swiggy or Zomato shows estimated delivery time and a friendly note.
"""
from flask import Flask, request, jsonify
import joblib
import threading

app = Flask(__name__)

# Load model
model = joblib.load("delivery_model.pkl")

@app.route("/predict-delivery", methods=["POST"])
def predict_delivery():

    data = request.get_json()

    distance = data["distance"]
    order_size = data["order_size"]

    prediction = model.predict([[distance, order_size]])[0]

    return jsonify({
        "predicted_delivery_time": round(prediction, 2),
        "message": "Your order is on the way! Estimated delivery time calculated successfully."
    })


# Run Flask in background
thread = threading.Thread(
    target=lambda: app.run(
        debug=False,
        use_reloader=False,
        port=5002
    )
)

thread.daemon = True
thread.start()

print("Delivery Prediction API started!")
print("URL: http://127.0.0.1:5002/predict-delivery")
```

```python
import requests

url = "http://127.0.0.1:5002/predict-delivery"

data = {
    "distance": 5,
    "order_size": 3
}

response = requests.post(url, json=data)

print(response.json())
```
