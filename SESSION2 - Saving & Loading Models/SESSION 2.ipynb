```python
"""
1.   Train a simple scikit-learn DecisionTreeClassifier on a small dataset of your choice (for example, a list of songs with
     genres from your Spotify playlist), then use Pickle to save the trained model to a file named song_genre_model.pkl.
"""

import pickle
from sklearn.tree import DecisionTreeClassifier

X = [
    [80, 30],
    [90, 40],
    [100, 50],
    [110, 60],
    [120, 70],
    [130, 80],
    [140, 90],
    [150, 95]
]

y = [
    0, 0, 0, 0,
    1, 1, 1, 1
]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

with open("song_genre_model.pkl", "wb") as file:
    pickle.dump(model, file)

print("Model trained successfully!")
print("Model saved as song_genre_model.pkl")
```

```python
"""
2.   Load the song_genre_model.pkl file you saved earlier and use it to predict the genre for a new song entry, printing the
     predicted result.
"""



import pickle

with open("song_genre_model.pkl", "rb") as file:
    model = pickle.load(file)

new_song = [[125, 75]]

prediction = model.predict(new_song)

genre = "Classical" if prediction[0] == 0 else "Pop"

print("Predicted Genre:", genre)
```

```python
"""
3.  Train a scikit-learn KNeighborsClassifier to predict whether a Flipkart product is 'electronics' or 'fashion' based on
    sample features, then use Joblib to save the model as product_category_model.joblib.<br><br><em><strong>Hint:</strong>
    Use joblib.dump() for saving and joblib.load() for loading.</em>
"""



import joblib
from sklearn.neighbors import KNeighborsClassifier

product_features = [
    [50000, 4.5],
    [45000, 4.2],
    [60000, 4.7],
    [1500, 3.8],
    [2000, 4.0],
    [2500, 4.3],
    [55000, 4.6],
    [1800, 4.1]
]

product_labels = [
    0, 0, 0,
    1, 1, 1,
    0, 1
]

product_model = KNeighborsClassifier(n_neighbors=3)
product_model.fit(product_features, product_labels)

joblib.dump(product_model, "product_category_model.joblib")

print("Product category model trained successfully!")
print("Model saved as product_category_model.joblib")
```
