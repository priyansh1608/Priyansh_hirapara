```python
#Q1
"""
1.  Install FastAPI and Uvicorn in a new Python virtual environment, then create a
   file called insta_like_api.py with a FastAPI app that exposes a root ('/') endpoint
   returning {'message': 'Instagram Like Counter API running'}.
"""
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Instagram Like Counter API running"}
```

```python
import uvicorn
import threading
import requests
import time

def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000)

thread = threading.Thread(target=run_server, daemon=True)
thread.start()

time.sleep(2)

response = requests.get("http://127.0.0.1:8000/")

print("Status Code:", response.status_code)
print("Response:", response.json())
```

```python
#Q2
"""
2.  Create a POST endpoint /predict-likes in your FastAPI app that accepts a JSON 
    payload with 'current_likes' (int) and 'new_likes' (int), and returns the updated
    total likes as JSON.
"""
from pydantic import BaseModel

class LikeRequest(BaseModel):
    current_likes: int
    new_likes: int

@app.post("/predict-likes")
def predict_likes(data: LikeRequest):
    total_likes = data.current_likes + data.new_likes
    
    return {
        "current_likes": data.current_likes,
        "new_likes": data.new_likes,
        "total_likes": total_likes
    }
```

```python
data = {
    "current_likes": 1200,
    "new_likes": 350
}

response = requests.post(
    "http://127.0.0.1:8000/predict-likes",
    json=data
)

print("Status Code:", response.status_code)
print("Response:", response.json())
```

```python
#Q4
"""
4.  Modify your FastAPI endpoint to handle the case where either 'current_likes' or 
   'new_likes' is missing in the request, and return an error message with status code
    400.<br><br><em><strong>Hint:</strong> Use Pydantic models and FastAPI's validation
    features to enforce required fields.</em>
"""
from fastapi import Request
from fastapi.exceptions import RequestValidationError
from fastapi.responses import JSONResponse

@app.exception_handler(RequestValidationError)
async def validation_exception_handler(request: Request, exc: RequestValidationError):
    return JSONResponse(
        status_code=400,
        content={
            "error": "current_likes and new_likes are required"
        }
    )
```

```python
data = {
    "current_likes": 1200
}

response = requests.post(
    "http://127.0.0.1:8000/predict-likes",
    json=data
)

print("Status Code:", response.status_code)
print("Response:", response.json())
```
